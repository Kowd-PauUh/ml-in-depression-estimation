```python
import os
from pathlib import Path

from tqdm.auto import tqdm

from src.audio_processing.waveform_operations import load_waveform
```

```python
def sizeof_fmt(num, suffix="B"):
    for unit in ("", "Ki", "Mi", "Gi", "Ti", "Pi", "Ei", "Zi"):
        if abs(num) < 1024.0:
            return f"{num:3.1f}{unit}{suffix}"
        num /= 1024.0
    return f"{num:.1f}Yi{suffix}"
```

```python
wav_dir = Path(os.environ['PROJECT_DIR']) / 'data' / 'raw_data'
wav_paths = [p.as_posix() for p in wav_dir.rglob('*.wav')]
```

```python
waveforms = {}
total_size = 0

pbar = tqdm(wav_paths, desc='Loading waveforms')
for path in pbar:
    waveform, sr = load_waveform(audio_path=path)
    waveforms[path] = waveform
    
    total_size += waveform.element_size() * waveform.nelement()
    pbar.set_postfix({'files': len(waveforms), 'size': sizeof_fmt(total_size)})
```
