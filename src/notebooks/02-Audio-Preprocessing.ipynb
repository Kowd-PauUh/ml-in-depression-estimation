```python
import os
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

from src.audio_processing.waveform_operations import load_waveform, trim_waveform
from src.audio_processing.waveform_augmentation import zero_mean
from src.audio_processing.features_extraction import mel_spectrogram
```

```python
MEL_BINS = 128
```

```python
df = pd.read_csv(Path(os.environ['PROJECT_DIR']) / 'data/preprocessed_data/data.csv')
```

```python
for _, row in tqdm(list(df.sample(100, random_state=42).iterrows()), desc='Computing MEL spectrograms'):
    waveform, sr = load_waveform(
        audio_path=row['source'],
        audio_format='wav'
    )
    waveform = trim_waveform(
        waveform=waveform, 
        sample_rate=sr, 
        start_time=row['start_time'],
        end_time=row['end_time'],
    )
    waveform = zero_mean(waveform=waveform)
    mel = mel_spectrogram(
        waveform=waveform,
        sample_rate=sr,
        num_mel_bins=MEL_BINS,
        length=3072,  # approximately 30s of audio
        padding=True,
        truncation=True
    )

    # plt.figure(figsize=(12, 2))
    # plt.imshow(mel.T)
```
